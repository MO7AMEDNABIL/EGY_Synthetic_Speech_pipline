```python
import torch
from transformers import pipeline

pipe = pipeline(
    "text-generation",
    model="MBZUAI-Paris/Nile-Chat-4B",
    model_kwargs={"torch_dtype": torch.bfloat16},
    device="cuda"
)
```

```python
prompt = """
You are generating training data for an Egyptian Arabic speech dataset.

Rules:
- Use ONLY spoken Egyptian Arabic (Cairo dialect)
- Do NOT use Modern Standard Arabic (MSA)
- Do NOT use formal Arabic
- Do NOT invent words
- Do NOT explain anything
- Do NOT number outputs
- Output ONLY 10 sentences, one per line

Examples:
عايز أطلب كشري
النت بطئ
هطلب اوبر
الجو حر النهارده
أنا رايح الشغل
فين أقرب صيدلية
محتاج أشحن الموبايل

Now generate 10 NEW sentences.
"""

messages = [
    {"role": "user", "content": prompt},
]

outputs = pipe(messages, max_new_tokens=500)
assistant_response = outputs[0]["generated_text"][-1]["content"].strip()
print(assistant_response)
```

```python
import re

def clean_llm_output(text):
    lines = text.split("\n")

    cleaned = []

    for line in lines:
        # remove numbering (١., 1., etc.)
        line = re.sub(r"^\s*[0-9٠-٩]+\s*[.\-]?\s*", "", line)

        line = line.strip()

        if len(line) > 3:
            cleaned.append(line)

    return cleaned
```

```python
llm_text = assistant_response  # your output

llm_prompts = clean_llm_output(llm_text)

print(llm_prompts)
```

```python
from flask import Flask, request, jsonify, send_file
from flask_cors import CORS
from pyngrok import ngrok
from TTS.api import TTS
import json
import os
import threading

app = Flask(__name__)

# CORS
CORS(app,
    resources={r"/*": {"origins": "*"}},
    expose_headers=["ngrok-skip-browser-warning"],
    allow_headers=[
        "Content-Type",
        "ngrok-skip-browser-warning"
    ]
)

tts = None
tts_lock = threading.Lock()

generation_status = {"running": False,"done": False,"error": None,"files": []}


def load_tts():
    global tts

    tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2")

def run_generation():
    global generation_status
    generation_status = {"running": True, "done": False,"error": None,"files": []}

    try:
        # Load prompts
        with open("data/prompts.json", encoding="utf-8") as f:
            prompts = json.load(f)

        # Create audio folder
        os.makedirs("data/audio", exist_ok=True)

        manifest = []

        speaker_wav = "speaker.wav"
        # Generate files
        for item in prompts:

            text = item["text"]

            audio_path = f"data/audio/{item['id']}.wav"

            print(f"Generating: {audio_path}")

            with tts_lock:
                tts.tts_to_file(text=text,
                    speaker_wav=speaker_wav,
                    language="ar",
                    file_path=audio_path
                )

            manifest.append({
                "id": item["id"],
                "text": text,
                "audio": audio_path
            })

            generation_status["files"].append({
                "id": item["id"],
                "text": text,
                "filename": f"{item['id']}.wav"
            })

            print(f"Generated: {audio_path}")

        # Save manifest
        with open("data/manifest.json","w",encoding="utf-8") as f:json.dump(manifest,f,ensure_ascii=False,indent=2)
        generation_status["running"] = False
        generation_status["done"] = True

    except Exception as e:
        generation_status["running"] = False
        generation_status["error"] = str(e)

        print("ERROR:", str(e))

@app.route("/generate", methods=["POST"])
def generate():

    if generation_status["running"]:
        return jsonify({"message": "Already running"}), 409

    t = threading.Thread(target=run_generation,daemon=True)
    t.start()

    return jsonify({"message": "Generation started"})

@app.route("/status", methods=["GET"])
def status():
    return jsonify(generation_status)

@app.route("/audio/<filename>", methods=["GET"])
def get_audio(filename):

    path = f"data/audio/{filename}"

    if not os.path.exists(path):
        return jsonify({"error": "File not found"}), 404

    response = send_file(path,mimetype="audio/wav")
    response.headers["ngrok-skip-browser-warning"] = "true" # help in ngrok compatibility to send audio to frontend
    return response


@app.route("/save", methods=["POST"]) # save data in HF format
def save():

    data = request.json
    os.makedirs("dataset/audio", exist_ok=True)
    metadata_path = "dataset/metadata.jsonl"

    with open(metadata_path, "w", encoding="utf-8") as f:

        for item in data:
            row = {
                "audio": f"audio/{item['id']}.wav",
                "text": item["text"]
            }

            f.write(
                json.dumps(row, ensure_ascii=False)
                + "\n"
            )

    return jsonify({"message": f"Saved {len(data)} entries in HuggingFace format." })
if __name__ == "__main__":
    print("Loading TTS model...")
    load_tts()
    print("TTS model loaded.")
    port = 5000
    public_url = ngrok.connect(port) # Create ngrok tunnel to connect with local frontend
    print("\n🌐 Public URL:")
    print(public_url)
    app.run(
        host="0.0.0.0",
        port=port,
        threaded=True
    )
```
